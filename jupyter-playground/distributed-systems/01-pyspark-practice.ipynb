Hay decenas de funciones útiles en `pyspark.sql.functions`. Las más usadas:

| Función | Para qué sirve |
|---------|----------------|
| `col(x)` | Referenciar una columna |
| `lit(v)` | Crear un valor constante |
| `when().otherwise()` | Lógica condicional |
| `expr("SQL_text")` | Escribir expresiones tipo SQL |
| `upper`, `lower`, `length` | Strings |
| `year`, `month`, `dayofmonth`, `to_date` | Fechas |
| `round`, `abs`, `ceil`, `floor` | Numéricas |
| `concat`, `concat_ws` | Concatenar strings |

**Enunciado:** Usando `df_ventas`, crea un nuevo DataFrame que tenga:
- Solo las ventas con `monto_total > 8.000`.
- Solo las columnas `id_venta`, `fecha`, `sucursal`, `monto_total`.
- Una columna nueva llamada `tipo_pago_completo` que sea `"En línea"` si el medio de pago es `"Crédito"` o `"Transferencia"`, y `"Físico"` en otros casos.
- Ordenado de mayor a menor monto.
- Mostrar solo las primeras 10 filas.

```python
solucion = (
    df_ventas
    .filter(col("monto_total") > 8000)
    .select("id_venta", "fecha", "sucursal", "monto_total", "medio_pago")
    .withColumn(
        "tipo_pago_completo",
        when(col("medio_pago").isin(["Crédito", "Transferencia"]), lit("En línea"))
        .otherwise(lit("Físico"))
    )
    .drop("medio_pago")
    .orderBy(col("monto_total").desc())
)
solucion.show(10)
```

**Enunciado**: A partir de `df_ventas`, crea un nuevo DataFrame que tenga:
- La columna `fecha` parseada como tipo `date` (no string).
- Una columna nueva `dia_semana` que tenga el número del día de la semana (1=domingo, 7=sábado).
- Una columna nueva `es_finde` que sea `True` si el día es sábado o domingo, `False` en caso contrario.
- Una columna nueva `año_mes` con el formato 'AAAA-MM' (ej: '2025-03').

Después, calcula:
1. ¿Qué porcentaje de las ventas ocurre en fin de semana?
2. ¿Cuál es el mes (`año_mes`) con mayor monto vendido?

```python
from pyspark.sql.functions import (
    col, to_date, dayofweek, when, lit,
    date_format, sum as _sum
)

df_temporal = (
    df_ventas
    .withColumn("fecha", to_date(col("fecha"), "yyyy-MM-dd"))
    .withColumn("dia_semana", dayofweek(col("fecha")))   # 1=domingo, 7=sábado
    .withColumn("es_finde",
        when(col("dia_semana").isin([1, 7]), lit(True)).otherwise(lit(False))
    )
    .withColumn("año_mes", date_format(col("fecha"), "yyyy-MM"))
)
df_temporal.select("fecha", "dia_semana", "es_finde", "año_mes", "monto_total").show(5)

# 1) Porcentaje de ventas en fin de semana
totales = df_temporal.agg(
    _sum("monto_total").alias("total"),
    _sum(when(col("es_finde"), col("monto_total")).otherwise(lit(0))).alias("total_finde"),
).first()
pct = (totales["total_finde"] / totales["total"]) * 100
print(f"1) Porcentaje en fin de semana: {pct:.1f}%")

# 2) Mes con mayor monto
df_temporal.groupBy("año_mes").agg(
    _sum("monto_total").alias("ventas_mes")
).orderBy(col("ventas_mes").desc()).show(1)
```

```python
# Leer las otras dos tablas
df_productos = spark.read.option("header", "true").option("inferSchema", "true").csv("01-data/productos.csv")
df_clientes = spark.read.option("header", "true").option("inferSchema", "true").csv("01-data/clientes.csv")

df_productos.show(3)
df_clientes.show(3)
```

```python
# INNER JOIN: solo filas que tengan match en ambas tablas
df_ventas_con_producto = df_ventas.join(
    df_productos,
    on="id_producto",       # columna por la que se une
    how="inner"             # tipo de join
)

df_ventas_con_producto.select(
    "id_venta", "fecha", "sucursal", "nombre", "categoria", "monto_total"
).show(5)
```

```python
# ¿Cuánto se vendió por categoría?
df_ventas_con_producto.groupBy("categoria").agg(
    count("*").alias("n_ventas"),
    _sum("monto_total").alias("ventas_totales"),
    avg("monto_total").alias("ticket_promedio")
).orderBy(col("ventas_totales").desc()).show()
```

```python
# Mejores 5 productos por monto vendido
df_ventas_con_producto.groupBy("nombre", "categoria").agg(
    _sum("monto_total").alias("ventas_totales"),
    _sum("cantidad").alias("unidades_vendidas")
).orderBy(col("ventas_totales").desc()).show(5, truncate=False)
```

```python
# Renombrar las columnas 'nombre' para que no choquen
df_productos_r = df_productos.withColumnRenamed("nombre", "nombre_producto")
df_clientes_r  = df_clientes.withColumnRenamed("nombre", "nombre_cliente")

# Triple join: ventas + productos + clientes
df_completo = (
    df_ventas
    .join(df_productos_r, on="id_producto", how="inner")
    .join(df_clientes_r,  on="id_cliente",  how="inner")
)

print("Columnas del DataFrame combinado:")
print(df_completo.columns)
df_completo.select(
    "id_venta", "nombre_producto", "categoria",
    "nombre_cliente", "comuna", "segmento", "monto_total"
).show(5)
```

```python
# Análisis: ¿qué segmento de cliente gasta más?
df_completo.groupBy("segmento").agg(
    countDistinct("id_cliente").alias("n_clientes"),
    count("*").alias("n_compras"),
    _sum("monto_total").alias("gasto_total"),
    avg("monto_total").alias("ticket_promedio"),
).orderBy(col("ticket_promedio").desc()).show()
```

```python
# Calculamos un resultado interesante
ranking_productos = (
    df_ventas_con_producto
    .groupBy("nombre", "categoria")
    .agg(_sum("monto_total").alias("ventas_totales"))
    .orderBy(col("ventas_totales").desc())
)

# Guardar como CSV (Spark escribe en una carpeta, no un solo archivo)
ranking_productos.write.mode("overwrite").option("header", "true").csv("01-data/output/resultados/ranking_csv")

# Guardar como Parquet (formato columnar, recomendado en Big Data)
ranking_productos.write.mode("overwrite").parquet("01-data/output/resultados/ranking_parquet")

print("Archivos escritos en la carpeta '01-data/output/resultados/'")
# !ls -la resultados/
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

# Definimos la "ventana": partir por sucursal, ordenar por monto desc
ventana = Window.partitionBy("sucursal").orderBy(col("monto_total").desc())

# Agregamos el ranking
df_ranking = df_ventas.withColumn("rank_sucursal", row_number().over(ventana))

# Top 3 ventas de cada sucursal
df_ranking.filter(col("rank_sucursal") <= 3).select(
    "sucursal", "rank_sucursal", "id_venta", "monto_total"
).orderBy("sucursal", "rank_sucursal").show(20)
```

```python
from pyspark.sql.functions import lag

# Otro ejemplo: lag para comparar venta con la anterior del mismo cliente
ventana_cliente = Window.partitionBy("id_cliente").orderBy("fecha")

df_lag = df_ventas.withColumn(
    "monto_anterior", lag(col("monto_total"), 1).over(ventana_cliente)
).withColumn(
    "diferencia", col("monto_total") - col("monto_anterior")
)

# Vemos solo los clientes que tienen al menos 2 compras
df_lag.filter(col("monto_anterior").isNotNull()).select(
    "id_cliente", "fecha", "monto_total", "monto_anterior", "diferencia"
).orderBy("id_cliente", "fecha").show(10)
```

**Caso integrador:**
1. KPIs globales del negocio
2. Top 3 sucursales por ventas
3. Top 3 productos más vendidos (por monto)
4. Distribución de ventas por segmento de cliente
5. Tendencia mensual de ventas
6. Guardar todo en archivos Parquet para el equipo de BI

```python
# Paso 1 — KPIs globales en una sola fila
from pyspark.sql.functions import sum as _sum, avg, count, countDistinct, col

kpis_globales = df_ventas.agg(
    _sum("monto_total").alias("ventas_totales_clp"),
    count("*").alias("n_transacciones"),
    countDistinct("id_cliente").alias("clientes_unicos"),
    countDistinct("id_producto").alias("productos_distintos"),
    avg("monto_total").alias("ticket_promedio"),
)
print("=== KPIs Globales ===")
kpis_globales.show(truncate=False)
```

```python
# Paso 2 — Top 3 sucursales por ventas
top_sucursales = (
    df_ventas.groupBy("sucursal")
    .agg(
        _sum("monto_total").alias("ventas_totales"),
        count("*").alias("n_ventas"),
        countDistinct("id_cliente").alias("clientes_unicos"),
    )
    .orderBy(col("ventas_totales").desc())
    .limit(3)
)
print("=== Top 3 Sucursales ===")
top_sucursales.show()
```

```python
# Paso 3 — Top 3 productos por monto vendido
top_productos = (
    df_completo.groupBy("nombre_producto", "categoria")
    .agg(
        _sum("monto_total").alias("ventas_totales"),
        _sum("cantidad").alias("unidades_vendidas"),
    )
    .orderBy(col("ventas_totales").desc())
    .limit(3)
)
print("=== Top 3 Productos ===")
top_productos.show(truncate=False)
```

```python
# Paso 4 — Distribución por segmento
distribucion_segmento = (
    df_completo.groupBy("segmento")
    .agg(
        countDistinct("id_cliente").alias("n_clientes"),
        _sum("monto_total").alias("gasto_total"),
        avg("monto_total").alias("ticket_promedio"),
    )
    .orderBy(col("gasto_total").desc())
)
print("=== Distribución por Segmento ===")
distribucion_segmento.show()
```

```python
# Paso 5 — Tendencia mensual
from pyspark.sql.functions import to_date, date_format

tendencia_mensual = (
    df_ventas
    .withColumn("año_mes", date_format(to_date(col("fecha")), "yyyy-MM"))
    .groupBy("año_mes")
    .agg(
        _sum("monto_total").alias("ventas_mes"),
        count("*").alias("n_ventas_mes"),
    )
    .orderBy("año_mes")
)
print("=== Tendencia Mensual ===")
tendencia_mensual.show()
```
